```python
# ============================================================
# FASHION PRODUCT ANALYSIS
# EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================


# ============================================================
# IMPORT LIBRARIES
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# UPLOAD DATASET
# ============================================================

from google.colab import files

uploaded = files.upload()

# Automatically get the uploaded filename

filename = list(uploaded.keys())[0]

print("\nUploaded File :", filename)


# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv(filename)


# ============================================================
# BASIC DATA UNDERSTANDING
# ============================================================

print("\nFirst 5 Rows:")

print(df.head())


print("\nDataset Shape:")

print(df.shape)


print("\nData Types:")

print(df.dtypes)


print("\nColumn Names:")

print(df.columns.tolist())


# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics:")

print(df.describe())


# ============================================================
# MISSING VALUES
# ============================================================

print("\nMissing Values:")

print(df.isnull().sum())


# ============================================================
# DUPLICATE ROWS
# ============================================================

print("\nDuplicate Rows:")

print(df.duplicated().sum())


# ============================================================
# HIGH PRICE PRODUCTS
# ============================================================

if "Price" in df.columns:

    high_price = df[
        df["Price"] > 300
    ]

    print("\nHigh Price Products:")

    print(high_price.head())


# ============================================================
# HIGH RATED PRODUCTS
# ============================================================

if "Rating" in df.columns:

    high_rating = df[
        df["Rating"] >= 4
    ]

    print("\nHighly Rated Products:")

    print(high_rating.head())


# ============================================================
# CATEGORY FILTER
# ============================================================

if "Category" in df.columns:

    most_common_category = (
        df["Category"].mode()[0]
    )

    category_products = df[
        df["Category"] == most_common_category
    ]

    print(
        "\nProducts in Most Common Category:",
        most_common_category
    )

    print(category_products.head())


# ============================================================
# SORT BY PRICE
# ============================================================

if "Price" in df.columns:

    sorted_df = df.sort_values(
        by="Price",
        ascending=False
    )

    print("\nMost Expensive Products:")

    print(sorted_df.head())


# ============================================================
# SORT BY RATING
# ============================================================

if "Rating" in df.columns:

    sorted_rating = df.sort_values(
        by="Rating",
        ascending=False
    )

    print("\nHighest Rated Products:")

    print(sorted_rating.head())


# ============================================================
# PRICE STATISTICS
# ============================================================

if "Price" in df.columns:

    print("\nPrice Statistics:")

    print(
        "Mean :",
        df["Price"].mean()
    )

    print(
        "Median :",
        df["Price"].median()
    )

    print("Mode :")

    print(
        df["Price"].mode()
    )

    print(
        "Standard Deviation :",
        df["Price"].std()
    )


# ============================================================
# RATING STATISTICS
# ============================================================

if "Rating" in df.columns:

    print("\nRating Statistics:")

    print(
        "Mean :",
        df["Rating"].mean()
    )

    print(
        "Median :",
        df["Rating"].median()
    )

    print("Mode :")

    print(
        df["Rating"].mode()
    )

    print(
        "Standard Deviation :",
        df["Rating"].std()
    )


# ============================================================
# CATEGORY COUNT
# ============================================================

if "Category" in df.columns:

    category_count = (
        df["Category"]
        .value_counts()
    )

    print("\nProducts by Category:")

    print(category_count)


# ============================================================
# BRAND COUNT
# ============================================================

if "Brand" in df.columns:

    brand_count = (
        df["Brand"]
        .value_counts()
    )

    print("\nProducts by Brand:")

    print(
        brand_count.head(10)
    )


# ============================================================
# AVERAGE PRICE BY CATEGORY
# ============================================================

if (
    "Category" in df.columns
    and "Price" in df.columns
):

    category_avg = (
        df.groupby("Category")["Price"]
        .mean()
    )

    print(
        "\nAverage Price by Category:"
    )

    print(
        category_avg
        .sort_values(ascending=False)
    )


# ============================================================
# TOP 10 CATEGORIES BY AVERAGE PRICE
# ============================================================

if (
    "Category" in df.columns
    and "Price" in df.columns
):

    category_avg.sort_values(
        ascending=False
    ).head(10).plot(
        kind="bar",
        figsize=(10,5)
    )

    plt.title(
        "Top 10 Categories by Average Price"
    )

    plt.xlabel("Category")

    plt.ylabel("Average Price")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# TOP 10 CATEGORIES BY PRODUCT COUNT
# ============================================================

if "Category" in df.columns:

    df["Category"].value_counts().head(10).plot(
        kind="bar",
        figsize=(10,5)
    )

    plt.title(
        "Top 10 Fashion Categories"
    )

    plt.xlabel("Category")

    plt.ylabel("Number of Products")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# TOP 10 BRANDS
# ============================================================

if "Brand" in df.columns:

    df["Brand"].value_counts().head(10).plot(
        kind="bar",
        figsize=(10,5)
    )

    plt.title(
        "Top 10 Fashion Brands"
    )

    plt.xlabel("Brand")

    plt.ylabel("Number of Products")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()


# ============================================================
# PRICE DISTRIBUTION
# ============================================================

if "Price" in df.columns:

    plt.figure(figsize=(8,5))

    sns.histplot(
        data=df,
        x="Price",
        bins=30,
        kde=True
    )

    plt.title(
        "Fashion Product Price Distribution"
    )

    plt.xlabel("Price")

    plt.ylabel("Number of Products")

    plt.show()


# ============================================================
# RATING DISTRIBUTION
# ============================================================

if "Rating" in df.columns:

    plt.figure(figsize=(8,5))

    sns.histplot(
        data=df,
        x="Rating",
        bins=20,
        kde=True
    )

    plt.title(
        "Fashion Product Rating Distribution"
    )

    plt.xlabel("Rating")

    plt.ylabel("Number of Products")

    plt.show()


# ============================================================
# PRICE BOXPLOT
# ============================================================

if "Price" in df.columns:

    plt.figure(figsize=(8,5))

    sns.boxplot(
        x=df["Price"]
    )

    plt.title(
        "Fashion Product Price Outliers"
    )

    plt.show()


# ============================================================
# CORRELATION
# ============================================================

numeric_df = df.select_dtypes(
    include="number"
)


if numeric_df.shape[1] >= 2:

    corr = numeric_df.corr()

    print("\nCorrelation Matrix:")

    print(corr)


# ============================================================
# CORRELATION HEATMAP
# ============================================================

if numeric_df.shape[1] >= 2:

    plt.figure(figsize=(10,7))

    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title(
        "Fashion Product Correlation Heatmap"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# PRICE VS RATING
# ============================================================

if (
    "Price" in df.columns
    and "Rating" in df.columns
):

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        data=df,
        x="Price",
        y="Rating"
    )

    plt.title(
        "Price vs Rating"
    )

    plt.xlabel("Price")

    plt.ylabel("Rating")

    plt.show()


# ============================================================
# PRICE VS RATING BY CATEGORY
# ============================================================

if (
    "Price" in df.columns
    and "Rating" in df.columns
    and "Category" in df.columns
):

    plt.figure(figsize=(10,6))

    sns.scatterplot(
        data=df,
        x="Price",
        y="Rating",
        hue="Category"
    )

    plt.title(
        "Price vs Rating by Category"
    )

    plt.xlabel("Price")

    plt.ylabel("Rating")

    plt.tight_layout()

    plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("==============================================")
print("       FASHION PRODUCT EDA COMPLETED")
print("==============================================")

print(
    "\nTotal Rows :",
    df.shape[0]
)

print(
    "Total Columns :",
    df.shape[1]
)

print(
    "\nNumeric Columns:"
)

print(
    df.select_dtypes(
        include="number"
    ).columns.tolist()
)

print(
    "\nCategorical Columns:"
)

print(
    df.select_dtypes(
        include="object"
    ).columns.tolist()
)

print(
    "\nEDA completed successfully!"
)
```
