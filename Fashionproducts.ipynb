```python
# ==========================================
# FASHION PRODUCTS DATA ANALYSIS PROJECT
# ==========================================

# Import Libraries
import pandas as pd
import numpy as np
import sqlite3
from google.colab import files

# ==========================================
# STEP 1: Upload CSV File
# ==========================================

print("Please upload fashion_products.csv")
uploaded = files.upload()

# Get uploaded filename automatically
filename = list(uploaded.keys())[0]

# Load dataset
df = pd.read_csv(filename)

print("\nDataset Loaded Successfully!")

# ==========================================
# Display Dataset
# ==========================================

print("\nFirst 5 Rows")
print(df.head())

print("\nLast 5 Rows")
print(df.tail())

# ==========================================
# Dataset Information
# ==========================================

print("\nDataset Information")
df.info()

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

# ==========================================
# Missing Values
# ==========================================

print("\nMissing Values")
print(df.isnull().sum())

# ==========================================
# Data Cleaning
# ==========================================

# Fill numeric missing values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Fill categorical missing values
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Cleaning")
print(df.isnull().sum())

# ==========================================
# Remove Duplicates
# ==========================================

print("\nDuplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates Removed")

# ==========================================
# Statistical Summary
# ==========================================

print("\nStatistical Summary")
print(df.describe(include='all'))

# ==========================================
# Export Files
# ==========================================

df.to_csv("FashionProducts_Cleaned.csv", index=False)
df.to_excel("FashionProducts_Cleaned.xlsx", index=False)
df.to_json("FashionProducts_Cleaned.json", orient="records", indent=4)

print("\nCSV, Excel and JSON exported successfully.")

# ==========================================
# Export to SQL
# ==========================================

conn = sqlite3.connect("FashionProducts.db")

df.to_sql(
    "fashion_products",
    conn,
    if_exists="replace",
    index=False
)

print("SQLite Database Created")

# ==========================================
# SQL - SELECT
# ==========================================

print("\nSELECT")
print(pd.read_sql("SELECT * FROM fashion_products LIMIT 5;", conn))

# ==========================================
# SQL - COUNT
# ==========================================

print("\nCOUNT")
print(pd.read_sql(
    "SELECT COUNT(*) AS Total_Products FROM fashion_products;",
    conn
))

# ==========================================
# SQL - WHERE
# ==========================================

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    query = f"""
    SELECT *
    FROM fashion_products
    WHERE "{col}" >
    (SELECT AVG("{col}") FROM fashion_products);
    """

    print("\nWHERE")
    print(pd.read_sql(query, conn))

# ==========================================
# SQL - ORDER BY
# ==========================================

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    query = f"""
    SELECT *
    FROM fashion_products
    ORDER BY "{col}" DESC
    LIMIT 10;
    """

    print("\nORDER BY")
    print(pd.read_sql(query, conn))

# ==========================================
# SQL - GROUP BY
# ==========================================

if len(categorical_cols) > 0:
    col = categorical_cols[0]

    query = f"""
    SELECT "{col}",
           COUNT(*) AS Total
    FROM fashion_products
    GROUP BY "{col}"
    ORDER BY Total DESC;
    """

    print("\nGROUP BY")
    print(pd.read_sql(query, conn))

# ==========================================
# SQL - AVG
# ==========================================

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    query = f"""
    SELECT AVG("{col}") AS Average_Value
    FROM fashion_products;
    """

    print("\nAVG")
    print(pd.read_sql(query, conn))

# ==========================================
# SQL - MAX
# ==========================================

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    query = f"""
    SELECT MAX("{col}") AS Maximum_Value
    FROM fashion_products;
    """

    print("\nMAX")
    print(pd.read_sql(query, conn))

# ==========================================
# SQL - MIN
# ==========================================

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    query = f"""
    SELECT MIN("{col}") AS Minimum_Value
    FROM fashion_products;
    """

    print("\nMIN")
    print(pd.read_sql(query, conn))

# ==========================================
# Close Database
# ==========================================

conn.close()

print("\nProject Completed Successfully!")
```
